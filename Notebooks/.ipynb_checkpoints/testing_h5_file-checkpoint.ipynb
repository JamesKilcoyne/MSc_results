## More rigourous checking of mapping: Ensure u-p map is ok

```python
with pd.option_context('display.max_rows', None,
    'display.max_columns', 8,
    'display.width', 13000,
    'display.precision', 10):

        
        print(df)
```

```python
#Give a iw and iz, and this will return the p values for the associated values fo a and e

def pvalue(iww,izz):
    pvals = []
    for _, row in df.iterrows():
        if row["iw"] == iww:
            if row["iz"]==izz:
                pvals.append(row["p"])

    return np.asarray(pvals)
```

```python
#for iu, iw = 0,3 the assoicated e,a is 0,0.9999982217

ALPHA_FLUX_nex=1/2

def p_of_u_flux_nex(u, pLSO):
    return (pLSO + DELTAPMIN_REGIONC) + (DELTAPMAX_REGIONC - DELTAPMIN_REGIONC) * (
        10**((u ** (1 / ALPHA_FLUX_nex) )* np.log10(2)) - 1  
    )

def u_of_p_nex2(p,pLSO):
    num = np.log10(p-pLSO) - np.log10(DELTAPMIN_REGIONC)
    den = np.log10(DELTAPMAX_REGIONC) - np.log10(DELTAPMIN_REGIONC)
    return num/den


def p_of_u_nex2(u,pLSO):
    return 10**(u*(np.log10(DELTAPMAX_REGIONC)-np.log10(DELTAPMIN_REGIONC)) + np.log10(DELTAPMIN_REGIONC))+pLSO


ee,aa,xx = 0.0,0.9999982217,1
plso = get_separatrix(aa,ee,xx)

uvals = np.linspace(0,1,8)

pvalss = p_of_u_flux_nex(uvals,plso)
pvalss2 = p_of_u_nex2(uvals,plso)

plt.plot(pvalss,".",label="p values based on my mapping")
plt.plot(pvalss2,".",label="p values based on my mapping")
plt.plot(pvalue(0,3),".",label="p values from the flux file")  
plt.grid()
plt.legend()
```
